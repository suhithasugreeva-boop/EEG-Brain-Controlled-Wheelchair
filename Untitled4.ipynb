```python
import mne

# 1. Load the EEG data
file_path = "/content/A01T[1].gdf"

raw = mne.io.read_raw_gdf(
    file_path,
    stim_channel="auto",
    preload=True
)

# Remove extra spaces from channel names
raw.rename_channels(lambda x: x.strip())

# 2. Extract events from annotations
events, event_id_mapping = mne.events_from_annotations(raw)

print("Available event codes:")
print(event_id_mapping)

# 3. Map original EEG classes to wheelchair commands
event_id = {
    "LEFT": int(event_id_mapping["769"]),
    "RIGHT": int(event_id_mapping["770"]),
    "FORWARD": int(event_id_mapping["771"]),
    "STOP": int(event_id_mapping["772"])
}

print("\nOur wheelchair event mapping:")
print(event_id)

# 4. Extract 4-second trials
epochs = mne.Epochs(
    raw,
    events,
    event_id=event_id,
    tmin=0.0,
    tmax=4.0,
    baseline=None,
    preload=True
)

print("\n--- Success! Epochs Created ---")
print(epochs)
```

```python
print("Number of trials:", len(epochs))
print("Epoch shape:", epochs.get_data().shape)
```

```python
import numpy as np

# Get EEG trial data
X = epochs.get_data()

# Get the original event labels
y = epochs.events[:, 2]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("First 20 original labels:", y[:20])
```

```python
import numpy as np

# EEG trial data
X = epochs.get_data()

print("X shape:", X.shape)
```

```python
label_map = {
    int(event_id_mapping['769']): 0,  # LEFT Hand
    int(event_id_mapping['770']): 1,  # RIGHT Hand
    int(event_id_mapping['771']): 2,  # FORWARD (Feet)
    int(event_id_mapping['772']): 3   # STOP (Tongue)
}

y = epochs.events[:, -1]

y_clean = np.array([label_map[label] for label in y])

print("Converted Labels (First 20):", y_clean[:20])
print("Unique labels in y_clean:", np.unique(y_clean))
```

```python
print("X shape:", X.shape)
print("y_clean shape:", y_clean.shape)
print("Unique labels:", np.unique(y_clean))
```

```python
# Find EEG channels and exclude EOG channels
eeg_picks = mne.pick_types(
    epochs.info,
    eeg=True,
    eog=False,
    stim=False
)

print("Number of EEG channels:", len(eeg_picks))
print("EEG channels:")
print([epochs.ch_names[i] for i in eeg_picks])
```

```python
X_eeg = X[:, eeg_picks, :]

print("Original X shape:", X.shape)
print("EEG-only X shape:", X_eeg.shape)
```

```python
# Apply 8–30 Hz band-pass filter to the EEG data

X_filtered = mne.filter.filter_data(
    X_eeg,
    sfreq=epochs.info["sfreq"],
    l_freq=8.0,
    h_freq=30.0,
    verbose=False
)

print("Filtered EEG shape:", X_filtered.shape)
```

```python
print("Minimum value:", X_filtered.min())
print("Maximum value:", X_filtered.max())
print("Mean value:", X_filtered.mean())
print("Standard deviation:", X_filtered.std())
```

```python
# Normalize the EEG data

mean = X_filtered.mean()
std = X_filtered.std()

X_normalized = (X_filtered - mean) / std

print("Normalized EEG shape:", X_normalized.shape)
print("Mean:", X_normalized.mean())
print("Standard deviation:", X_normalized.std())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_normalized,
    y_clean,
    test_size=0.2,
    random_state=42,
    stratify=y_clean
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Change shape from:
# (trials, channels, time)
# to:
# (trials, time, channels)

X_train = np.transpose(X_train, (0, 2, 1))
X_test = np.transpose(X_test, (0, 2, 1))

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
print("Training labels:", np.unique(y_train))
print("Testing labels:", np.unique(y_test))

print("\nTraining samples per class:")
print(np.bincount(y_train))

print("\nTesting samples per class:")
print(np.bincount(y_test))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM, Dense, Dropout
```

```python
model = Sequential([

    # CNN layer
    Conv1D(
        filters=32,
        kernel_size=5,
        activation="relu",
        input_shape=(X_train.shape[1], X_train.shape[2])
    ),

    # Reduce the time dimension
    MaxPooling1D(pool_size=2),

    # Prevent overfitting
    Dropout(0.3),

    # LSTM layer
    LSTM(64),

    # Another dropout layer
    Dropout(0.3),

    # Fully connected layer
    Dense(32, activation="relu"),

    # Final 4-class output
    Dense(4, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("CNN-LSTM model compiled successfully!")
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("CNN-LSTM Training and Validation Accuracy")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("CNN-LSTM Training and Validation Loss")
plt.legend()
plt.show()
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print("Test Accuracy (%):", test_accuracy * 100)
```

```python
y_pred_prob = model.predict(X_test)

# Convert probabilities into predicted class numbers
y_pred = np.argmax(y_pred_prob, axis=1)

print("Predicted labels:", y_pred[:20])
print("Actual labels:   ", y_test[:20])
```

```python
from sklearn.metrics import classification_report

class_names = [
    "LEFT",
    "RIGHT",
    "FORWARD",
    "STOP"
]

print(
    classification_report(
        y_test,
        y_pred,
        target_names=class_names
    )
)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot()
plt.title("CNN-LSTM Confusion Matrix")
plt.show()
```

```python
label_map_eegnet = {
    event_id_mapping["769"]: 0,  # LEFT
    event_id_mapping["770"]: 1,  # RIGHT
    event_id_mapping["771"]: 2,  # FORWARD
    event_id_mapping["772"]: 3   # STOP
}

y_eegnet = np.array([
    label_map_eegnet[event]
    for event in epochs_eegnet.events[:, -1]
])

print("Labels shape:", y_eegnet.shape)
print("Class counts:", np.bincount(y_eegnet))
```

```python
from sklearn.model_selection import train_test_split

X_train_eegnet, X_test_eegnet, y_train_eegnet, y_test_eegnet = train_test_split(
    X_eegnet,
    y_eegnet,
    test_size=0.2,
    random_state=42,
    stratify=y_eegnet
)

print("Training data:", X_train_eegnet.shape)
print("Testing data:", X_test_eegnet.shape)
print("Training labels:", y_train_eegnet.shape)
print("Testing labels:", y_test_eegnet.shape)
```

```python
train_mean = X_train_eegnet.mean(axis=(0, 2), keepdims=True)
train_std = X_train_eegnet.std(axis=(0, 2), keepdims=True)

X_train_eegnet = (
    X_train_eegnet - train_mean
) / (train_std + 1e-8)

X_test_eegnet = (
    X_test_eegnet - train_mean
) / (train_std + 1e-8)

print("Training mean:", X_train_eegnet.mean())
print("Training std:", X_train_eegnet.std())
```

```python
X_train_eegnet = X_train_eegnet[..., np.newaxis]
X_test_eegnet = X_test_eegnet[..., np.newaxis]

print("EEGNet training shape:", X_train_eegnet.shape)
print("EEGNet testing shape:", X_test_eegnet.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, DepthwiseConv2D, BatchNormalization,
    Activation, AveragePooling2D, Dropout, Flatten, Dense
)

# EEGNet input
inputs = Input(shape=(22, 751, 1))

# Block 1: Temporal filtering
x = Conv2D(
    filters=8,
    kernel_size=(1, 64),
    padding="same",
    use_bias=False
)(inputs)

x = BatchNormalization()(x)

# Block 2: Spatial filtering across EEG channels
x = DepthwiseConv2D(
    kernel_size=(22, 1),
    depth_multiplier=2,
    padding="valid",
    use_bias=False
)(x)

x = BatchNormalization()(x)
x = Activation("elu")(x)

x = AveragePooling2D(pool_size=(1, 4))(x)
x = Dropout(0.5)(x)

# Block 3: Separable temporal filtering
x = tf.keras.layers.SeparableConv2D(
    filters=16,
    kernel_size=(1, 16),
    padding="same",
    use_bias=False
)(x)

x = BatchNormalization()(x)
x = Activation("elu")(x)

x = AveragePooling2D(pool_size=(1, 8))(x)
x = Dropout(0.5)(x)

# Classification layer
x = Flatten()(x)

outputs = Dense(
    4,
    activation="softmax"
)(x)

eegnet_model = Model(inputs=inputs, outputs=outputs)

eegnet_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

eegnet_model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history_eegnet = eegnet_model.fit(
    X_train_eegnet,
    y_train_eegnet,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
test_loss, test_accuracy = eegnet_model.evaluate(
    X_test_eegnet,
    y_test_eegnet,
    verbose=1
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np

y_pred_prob = eegnet_model.predict(X_test_eegnet)
y_pred = np.argmax(y_pred_prob, axis=1)

class_names = ["LEFT", "RIGHT", "FORWARD", "STOP"]

cm = confusion_matrix(y_test_eegnet, y_pred)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test_eegnet,
        y_pred,
        target_names=class_names,
        zero_division=0
    )
)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization, DepthwiseConv2D,
    SeparableConv2D, Activation, AveragePooling2D,
    Dropout, Flatten, Dense
)
from tensorflow.keras.regularizers import l2

# Input shape
inputs = Input(shape=(22, 751, 1))

# Temporal convolution
x = Conv2D(
    filters=8,
    kernel_size=(1, 64),
    padding="same",
    use_bias=False,
    kernel_regularizer=l2(1e-4)
)(inputs)

x = BatchNormalization()(x)

# Spatial convolution across all EEG channels
x = DepthwiseConv2D(
    kernel_size=(22, 1),
    depth_multiplier=2,
    padding="valid",
    use_bias=False,
    depthwise_regularizer=l2(1e-4)
)(x)

x = BatchNormalization()(x)
x = Activation("elu")(x)

x = AveragePooling2D(pool_size=(1, 4))(x)
x = Dropout(0.5)(x)

# Separable convolution
x = SeparableConv2D(
    filters=16,
    kernel_size=(1, 16),
    padding="same",
    use_bias=False,
    depthwise_regularizer=l2(1e-4),
    pointwise_regularizer=l2(1e-4)
)(x)

x = BatchNormalization()(x)
x = Activation("elu")(x)

x = AveragePooling2D(pool_size=(1, 8))(x)
x = Dropout(0.6)(x)

# Classification
x = Flatten()(x)

outputs = Dense(
    4,
    activation="softmax",
    kernel_regularizer=l2(1e-4)
)(x)

improved_eegnet = Model(inputs=inputs, outputs=outputs)

improved_eegnet.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

improved_eegnet.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization,
    DepthwiseConv2D, SeparableConv2D,
    Activation, AveragePooling2D,
    Dropout, Flatten, Dense
)

# EEGNet parameters
F1 = 8
D = 2
F2 = 16
dropout_rate = 0.5

# Input: 22 EEG channels × 751 time samples × 1
inputs = Input(shape=(22, 751, 1))

# ------------------------------------------------
# Block 1: Temporal convolution
# ------------------------------------------------
x = Conv2D(
    F1,
    kernel_size=(1, 64),
    padding="same",
    use_bias=False
)(inputs)

x = BatchNormalization()(x)

# ------------------------------------------------
# Block 2: Depthwise spatial convolution
# ------------------------------------------------
x = DepthwiseConv2D(
    kernel_size=(22, 1),
    depth_multiplier=D,
    padding="valid",
    use_bias=False
)(x)

x = BatchNormalization()(x)
x = Activation("elu")(x)

x = AveragePooling2D(pool_size=(1, 4))(x)
x = Dropout(dropout_rate)(x)

# ------------------------------------------------
# Block 3: Separable convolution
# ------------------------------------------------
x = SeparableConv2D(
    F2,
    kernel_size=(1, 16),
    padding="same",
    use_bias=False
)(x)

x = BatchNormalization()(x)
x = Activation("elu")(x)

x = AveragePooling2D(pool_size=(1, 8))(x)
x = Dropout(dropout_rate)(x)

# ------------------------------------------------
# Classification
# ------------------------------------------------
x = Flatten()(x)

outputs = Dense(
    4,
    activation="softmax"
)(x)

eegnet = Model(inputs=inputs, outputs=outputs)

eegnet.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

eegnet.summary()
```
